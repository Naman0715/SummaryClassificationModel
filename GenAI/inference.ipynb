## Import Required Libraries
This cell imports the necessary components from the `transformers` library:
- **AutoTokenizer**: Automatically loads the correct tokenizer for the model
- **AutoModelForSeq2SeqLM**: Loads a pre-trained sequence-to-sequence model for text summarization

```python
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM
```

## Load Pre-trained Model and Tokenizer
This cell loads the T5 model and its corresponding tokenizer from the local `./model` directory. The tokenizer is responsible for converting raw text into token IDs that the model can process.

```python
model_path = "./model"

# Load model

tokenizer = AutoTokenizer.from_pretrained(model_path)
model = AutoModelForSeq2SeqLM.from_pretrained(model_path)
```

## Sample Input Text
Here we define a sample conversation between Alice and Bob that will be used for summarization testing.

```python
# Test text

text = """
Alice: Hey did you finish the project?
Bob: Almost done.
Alice: The quickly come to my office.
"""
```

## Tokenize Input Text
This cell converts the raw text into tokenized format that the model expects. The `return_tensors="pt"` parameter ensures the output is in PyTorch tensor format. The resulting tokens and attention masks are printed for verification.

```python
# Tokenize

inputs = tokenizer(text, return_tensors="pt")
print(inputs)
```

## Generate Summary
This cell uses the model to generate a summary from the input text. The `max_new_tokens=50` parameter limits the summary to a maximum of 50 tokens, ensuring conciseness.

```python
# Generate summary

outputs = model.generate(
    **inputs,
    max_new_tokens=50
)

print(outputs)
```

## Decode and Display Results
This cell decodes the generated token IDs back into human-readable text. The `skip_special_tokens=True` parameter removes special tokens from the output, showing only the actual summary text.

```python
summary = tokenizer.decode(outputs[0], skip_special_tokens=True)
```

```python
print("\nGenerated Summary:\n")
print(summary)
```
